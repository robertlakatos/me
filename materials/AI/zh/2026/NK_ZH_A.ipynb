#### A Kő-Papír-Olló Játék Működése és a Q-learning Ügynök Tanulása (Csak a teszteléshez)

Ebben a tesztalkalmazásban egy Q-learning ügynök próbál megtanulni nyerni a kő-papír-olló játékban egy speciális ellenfél ellen. A játék jellemzői a következők:

*   **Állapot (State):** Mivel a kő-papír-olló egy fordulatos játék, ahol az előző lépések nem befolyásolják a jövőbeli optimális döntést (ellentétben például egy sakkjátékkal), ezért az ügynök számára egyetlen, fix állapotot (0,0) definiálunk.
*   **Akciók (Actions):** Három lehetséges akció van: 0 (Kő), 1 (Papír), 2 (Olló).
*   **Ellenfél Stratégiája:** Az ellenfél stratégiája nem teljesen véletlenszerű, hanem eltolt:
    *   80%-ban Papírt (1) választ.
    *   10%-ban Követ (0) választ.
    *   10%-ban Ollót (2) választ.
*   **Jutalom (Reward):** A jutalmazási rendszer a következő:
    *   **1 pont:** Ha az ügynök nyer (pl. az ügynök Ollót, az ellenfél Papírt választ).
    *   **0 pont:** Döntetlen esetén.
    *   **-1 pont:** Ha az ügynök veszít.

```python
n_states_row = 1  # Single state for the RPS game
n_states_cols = 1 # Single state for the RPS game
n_actions = 3     # 0: Rock, 1: Paper, 2: Scissors
learning_rate = 0.1

agent = QLearningAgent(n_states_row, n_states_cols, n_actions, learning_rate)

epsilon = 0.1 # Epsilon remains constant
gamma = 0.99  # Gamma starts high and will decrease

print("--- Q-Learning Agent RPS Test Application ---")

num_episodes = 200 # Increased episodes for learning

gamma_decay_rate = 0.002 # Slower decay for more learning
min_gamma = 0.1

# Helper function for opponent's move
def get_opponent_move():
    # 80% Paper (1), 10% Rock (0), 10% Scissors (2)
    choice = random.uniform(0, 1)
    if choice < 0.10: # 10%
        return 0 # Rock
    elif choice < 0.90: # 80%
        return 1 # Paper
    else: # 10%
        return 2 # Scissors

# Helper function for reward calculation
def get_reward(agent_action, opponent_move):
    # 0: Rock, 1: Paper, 2: Scissors
    if agent_action == opponent_move:
        return 0 # Draw
    elif (agent_action == 0 and opponent_move == 2) or \
         (agent_action == 1 and opponent_move == 0) or \
         (agent_action == 2 and opponent_move == 1):
        return 1 # Agent wins
    else:
        return -1 # Agent loses

# Action names for printing
action_names = {0: "Rock", 1: "Paper", 2: "Scissors"}

for episode in range(num_episodes):
    # In RPS, the "state" doesn't change per round, so we use a fixed state (0,0)
    current_state_row = 0
    current_state_col = 0

    if (episode + 1) % 20 == 0 or episode == 0: # Print less frequently for brevity
        print(f"\nEpisode {episode + 1} - Gamma: {gamma:.4f}")
        print(f"  Q-values for state (0,0): {agent.q_table[0, 0, :]}")

    # Simulate a few steps within an episode (e.g., a few rounds of RPS)
    for step in range(5):
        # Agent chooses an action
        agent_action = agent.act(current_state_row, current_state_col, epsilon)

        # Opponent makes a move
        opponent_move = get_opponent_move()

        # Calculate reward
        reward = get_reward(agent_action, opponent_move)

        # New state is the same as current state for this game
        new_state_row = 0
        new_state_col = 0

        if (episode + 1) % 20 == 0 and step == 0: # Print only for the first step of selected episodes
             print(f"  Agent plays: {action_names[agent_action]}, Opponent plays: {action_names[opponent_move]}, Reward: {reward}")

        # Agent learns from the experience
        agent.learn(current_state_row, current_state_col, agent_action, reward, new_state_row, new_state_col, gamma)

    # Decrease gamma for the next episode
    gamma = max(min_gamma, gamma - gamma_decay_rate)

print("\n--- Test Finished ---")
print("Final Q-table for state (0,0):")
print(agent.q_table[0, 0, :])
print("Final Gamma:", gamma)
print("Interpretation: The agent should learn that playing Scissors (index 2) yields the highest Q-value against an opponent that mostly plays Paper (index 1).")
```
